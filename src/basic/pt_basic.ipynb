## Introduction to Tensor

```python
# Scalar
t0 = torch.tensor(7.)
t0,t0.dtype,t0.ndim,t0.item()
```

```python
# Vector
t1 = torch.tensor([1., 2.])
t1,t1.dtype,t1.ndim,t1.shape
```

```python
# Matrix
M1 = torch.tensor([[1., 2.],
                   [3., 4.]])
M1,M1.dtype,M1.ndim,M1.shape
```

```python
# Random
T = torch.randn(size=[3, 4, 3])
T,T.dtype,T.ndim,T.shape
```

```python
# Zeros
T = torch.zeros(size=[4, 3])
T,T.dtype,T.ndim,T.shape
```

```python
# Range
T = torch.arange(0, 10)
T
```

```python
# Like
T = torch.ones_like(input=T, dtype=torch.float16)
T,T.dtype
```

```python
# Convert
T = T.type(torch.float32)
T.dtype,T.device
```

## Manipulating Tensors

```python
# Elementwise manipulations (+, -, *, /)
T = torch.tensor([1, 2, 3])
(T + 10) * 2 - 1 / 2
```

```python
# Matrix multiplication
T = torch.tensor([[1, 2, 3],
                  [4, 5, 6]])
T = torch.tensor([[1, 2],
                  [3, 4],
                  [5, 6]])
T = torch.matmul(T, T) # 'T1 @ T2' and torch.mm variants are the same
T
```

```python
# Transpose
T.T
```

```python
# Min / Max / Mean
T = torch.rand(size=[3, 4, 5], dtype=torch.float32)
T.min(), T.max(), T.mean()
```

```python
# Finding the min max position
T = torch.arange(1, 100)
torch.argmin(T), torch.argmax(T), T[torch.argmax(T)]
```

```python
# Reshape (reshapes an input tensor to a defined shape)
T = torch.arange(1., 11.)
T = T.reshape(shape=(2, 5))
T
```

```python
# View (return a view of an input tensor of certain shape but keep the same memory)
T = torch.arange(1., 11.)
Y = T.view(size=(10, 1))
Y[0,0] = 99
T
```

```python
# Stacking (combine multiple tensors on top or side by side of each other)
T = torch.tensor([1, 2, 3])
Y = torch.stack([T, T, T], dim=1)
Y
```

```python
# Squeeze / Unsqueeze (remove all 1 dimensions from a tensor)
T = torch.rand(size=[2, 1, 3])
Y = torch.squeeze(T)
Z = torch.unsqueeze(Y, dim=0)
T.shape,Y.shape,Z.shape
```

```python
# Permute (return a view of the input with dimensions swapped in a certain way)
# (B, W, H, C) => (B, C, W, H)
T = torch.rand(size=[1, 224, 224, 3])
Y = torch.permute(T, (0, 3, 2, 1))
T.shape,Y.shape
```

```python
SEED = 13
torch.manual_seed(SEED)
T1 = torch.rand(3, 4)
torch.manual_seed(SEED)
T2 = torch.rand(3, 4)
print(T1 == T2)
```
